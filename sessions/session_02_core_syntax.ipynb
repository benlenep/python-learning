## 2. Strings and f-strings

In R a function takes the string (`toupper(s)`). In Python the string carries its own **methods** (`s.upper()`).

`paste()`, `sprintf()` and `glue()` all become **f-strings**: put `f` before the quotes and expressions inside `{}`.

````python
team = "geelong cats"

print(team.upper())                    # toupper(team)
print(team.title())                    # tools::toTitleCase(team), roughly
print(team.replace("cats", "Cats"))    # sub("cats", "Cats", team)
print(len(team))                       # nchar(team)
print("cats" in team)                  # grepl("cats", team, fixed = TRUE)
print(team.split(" "))                 # strsplit(team, " ")[[1]]
````

````python
members = 1234567
rate = 0.08734

print(f"Members: {members:,}")          # thousands separator
print(f"Rate: {rate:.1%}")              # percentage, 1 decimal place
print(f"Rate: {rate:.3f}")              # 3 decimal places
print(f"{'left':<10}|{'right':>10}")    # pad and align within 10 characters
````

## 3. Lists versus R vectors

This section holds the five traps that catch nearly every R user:

1. **Indexing starts at 0.** `x[0]` is the first element.
2. **Slices exclude the end.** `x[1:3]` gives the 2nd and 3rd elements, not three elements.
3. **Negative indices count from the end.** `x[-1]` is the *last* element. In R, `x[-1]` *drops the first*. Same syntax, opposite meaning.
4. **Lists are not vectorised.** `[1, 2] * 2` repeats the list. Element-wise maths needs a comprehension (section 5) or numpy and pandas (Phase 2).
5. **Assignment does not copy.** `b = a` gives one list two names. R copies on modify; Python does not.

````python
quarters = [3, 5, 4, 3]    # goals by quarter

print(quarters[0])      # first element: 3       (R: quarters[1])
print(quarters[-1])     # last element: 3        (R: quarters[length(quarters)])
print(quarters[1:3])    # [5, 4]                 (R: quarters[2:3])
print(quarters[:2])     # first two              (R: head(quarters, 2))
print(len(quarters), sum(quarters), max(quarters))
````

````python
print([1, 2] * 2)       # [1, 2, 1, 2]  repetition, not multiplication
print([1, 2] + [3])     # [1, 2, 3]     concatenation, like c(c(1, 2), 3)

quarters.append(2)      # modifies the list in place and returns None
print(quarters)
````

````python
a = [1, 2, 3]
b = a              # NOT a copy: a and b are the same list
b.append(99)
print("a after changing b:", a)

c = a.copy()       # a genuine copy
c.append(0)
print("a:", a, " c:", c)
````

## 4. Dictionaries versus named lists

A dictionary maps unique **keys** to values, much like a named list in R: `list(name = "Geelong", founded = 1859)` becomes `{"name": "Geelong", "founded": 1859}`.

| R | Python |
| --- | --- |
| `club$name` or `club[["name"]]` | `club["name"]` |
| `names(club)` | `club.keys()` |
| `club$colours <- "navy and white"` | `club["colours"] = "navy and white"` |
| `"name" %in% names(club)` | `"name" in club` |

````python
club = {"name": "Geelong", "ground": "Kardinia Park", "founded": 1859}

print(club["name"])
print(club.get("mascot", "unknown"))   # safe lookup with a default; club["mascot"] would raise KeyError

club["colours"] = "navy and white"     # add a new key
print(list(club.keys()))

for key, value in club.items():        # loop over key and value together
    print(f"{key}: {value}")
````

## 5. Loops and comprehensions

- **Indentation is syntax.** The indented block *is* the loop body; there are no braces.
- `range(1, 5)` gives 1, 2, 3, 4. The end is excluded, as with slices.
- `enumerate()` gives you the index and the item together; `zip()` walks two lists side by side.
- A **list comprehension** is Python's answer to `sapply()` and `purrr::map()`. Read `[s * 2 for s in scores if s > 50]` as "s times 2, for each s in scores, keeping those over 50".

````python
for q in range(1, 5):
    print(f"Quarter {q}")

teams = ["Geelong", "Sydney", "Brisbane"]

for i, t in enumerate(teams, start=1):    # seq_along(teams) plus the value
    print(i, t)

for t, pts in zip(teams, [60, 64, 52]):    # walk two lists in step
    print(t, pts)
````

````python
goals = [15, 9, 12]
behinds = [12, 8, 10]

scores = [g * 6 + b for g, b in zip(goals, behinds)]    # purrr::map2_dbl(goals, behinds, ~ .x * 6 + .y)
print(scores)

big = [s for s in scores if s >= 80]                     # purrr::keep(scores, ~ .x >= 80)
print(big)

score_by_team = {t: s for t, s in zip(teams, scores)}    # setNames(scores, teams), as a dict
print(score_by_team)
````
